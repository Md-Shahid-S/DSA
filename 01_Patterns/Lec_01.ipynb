```python
n = 4
total_width = 2 * n - 1  # 7 spaces allocated for the left half + gap

for i in range(1, n + 1):
    # Left string: 1 to i
    left = ''.join(str(j) for j in range(1, i + 1))
    
    # Right string: i down to 1
    right = ''.join(str(j) for j in range(i, 0, -1))
    
    # Left side gets padded with spaces to perfectly align columns
    print(left.ljust(total_width) + right)
```

```python
n = 4
for i in range(1, n + 1):
    # left: 1 to i
    left = ''
    for j in range(1, i + 1):
        left += str(j)

    # right: i down to 1
    right = ''
    for j in range(i, 0, -1):
        right += str(j)

    # To match the image, row 1 needs 6 spaces, row 2 needs 4, row 3 needs 2
    spaces = ' ' * (n - i) * 2

    print(left + spaces + right)
```

```python
n = 5

# top half — stars shrink, gap grows
for i in range(1, n + 1):
    stars = n - i + 1
    gap   = 2 * (i - 1)
    print('* ' * stars + ' ' * gap + '* ' * stars)

# bottom half — stars grow, gap shrinks (start from 2 to avoid duplicate)
for i in range(n, 0, -1):
    stars = i
    gap   = 2 * (n - i)
    print('* ' * stars + ' ' * gap + '* ' * stars)
```

```python
n = 4
size = 2 * n - 1  # 7x7 grid

for i in range(size):
    for j in range(size):
        # min distance from any of the 4 edges
        dist = min(i, j, size - 1 - i, size - 1 - j)
        print(n - dist, end=' ')
    print()
```
