```python
for i in range(0, 6):
    for j in range(0, 6):
        if i == 0 or j == 0  or i == 5 or j == 5:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
for i in range(0,5):
    for j in range(0,5):
        if i+j <= 4:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
for i in range(0,5):
    for j in range(0,5):
        if i+j >= 4:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
for i in range(1, 6):          # Controls the number of rows (1 to 5)
    for j in range(1, i + 1):  # Controls the numbers printed in each row
        print(j, end=' ')       # Prints the number without a newline
    print()                    # Moves to the next line after the inner loop finishes
```

```python
for i in range(1, 6):          # Outer loop: Row numbers 1, 2, 3, 4
    for j in range(1, i + 1):  # Inner loop: Runs 'i' times
        print(i, end=' ')       # Print the current row number
    print()                    # New line after each row
```

```python
for i in range(5, 0, -1):          # Outer loop: Row numbers 5, 4, 3, 2, 1
    for j in range(1, i + 1): 
        print(j, end=' ')      
    print()                    
```

```python
for i in range(5, 0, -1):          # Outer loop: Row numbers 6, 5, 4, 3, 2
    for j in range(1, i + 1): 
        print(i, end=' ')      
    print()                    
```

```python
for i in range(1,10):
    for j in range(1,10):
        print(i,j, end=' \t')
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    # 1. Print leading spaces
    for j in range(rows - i):
        print(' ', end=' ')
    
    # 2. Print stars
    for k in range(2 * i - 1):
        print('*', end=' ')
    
    # 3. Move to next line
    print()
```

```python
rows = 5
for i in range(rows , 0, -1):
    # 1. Print leading spaces
    for j in range(rows - i):
        print(' ', end=' ')
    
    # 2. Print stars
    for k in range(2 * i - 1):
        print('*', end=' ')
    
    # 3. Move to next line
    print()
```

```python
n = 5
for i in range(1,n+1):
    for j in range(1, n-i-1):
        print(' ', end=' ')
    for k in range(2*i-1):
        print('*', end=' ')
    for l in range(1, n-i-1):
        print(' ', end=' ')
        
    print()
```

```python
n = 5
for i in range(1, n + 1):
    for j in range(n - i):      # leading spaces
        print(' ', end=' ')
    for k in range(2 * i - 1):  # stars
        print('*', end=' ')
    print()                     # newline
```
