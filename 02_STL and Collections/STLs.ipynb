```python
from collections import deque, Counter, defaultdict
import heapq
import itertools
```

```python
# vector → list
arr = [1, 2, 3]
arr.append(4)
arr
```

```python
# stack → list
stack = []
stack.append(1)   # push
print(stack)
stack.pop()       # pop
print(stack)
```

```python
# queue → deque
q = deque()
q.append(1)       # enqueue
print(q)
q.popleft()       # dequeue
print(q)
```

```python
# deque
dq = deque()
dq.appendleft(1)  # insert front
print(dq)
dq.append(2)      # insert back
print(dq)
```

```python
# priority_queue → heapq (min heap)
h = []
heapq.heappush(h, 3)
heapq.heappush(h, 1)
print(heapq.heappop(h))   # 1 (smallest)
```

```python
# max heap → negate values
heapq.heappush(h, -5)
print(-heapq.heappop(h))  # 5 (largest)
```

```python
# set
s = {1, 2, 3, 2}   # {1, 2, 3} — duplicates removed
print(s)
```

```python
# multiset (duplicates allowed)
from collections import Counter
c = Counter([1, 2, 2, 3, 3, 3])  # {3:3, 2:2, 1:1}
print(c)
```

```python
# map → dict
d = {'a': 1, 'b': 2}
print(d['a'])  # 1
```

```python
# multimap → defaultdict
mm = defaultdict(list)
mm['a'].append(1)
mm['a'].append(2)   # {'a': [1, 2]}
print(mm['a'])
```

```python
# sort
arr = [3, 1, 2]
arr.sort()                        # in-place
print(arr)
arr = sorted(arr, reverse=True)   # descending
print(arr)
print()

# min / max
print('Min and Max:')
print(min(arr), max(arr))
```

```python
# next_permutation
for p in itertools.permutations([1, 2, 3]):
    print(p)
    
```

```python
# popcount (count 1-bits)
n = 7   # binary: 111
print(bin(n).count('1'))   # 3
```
